```python
from datetime import datetime
def new_salary(cargo, salary, ano_adm):
    ano_atual = datetime.now().year
    if cargo.upper()  == 'GERENTE':
        perc = 1.08
    elif cargo.upper() == 'ENGENHEIRO':
        perc = 1.09

    elif cargo.upper() == 'TECNICO':
        perc = 1.1
    else:
        return  salary *1.07
    
    tp_trabalho = ano_atual - int(ano_adm)

    if tp_trabalho < 3:
        add = 0.01
    elif tp_trabalho >=3 and tp_trabalho <5:
        add = 0.01
    else:
        add = 0.02

    perc_total = perc + add

    return perc_total * salary

cargo = input('Digite o cargo do funcionário:')
salario = int(input('Digite o salário do funcionário:'))
ano_adm = int(input('Digite o ano adimição do funcionário:'))
resp = new_salary(cargo, salario, ano_adm)

print(f'O salário antigo era de:{salario}\nO novo salário do funcionário será de {resp}\nA diferença é de {resp-salario}')
```
