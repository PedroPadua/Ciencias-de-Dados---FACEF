```python
import pandas as pd

df = pd.read_csv('dados_malaria.csv')
```

```python
# Exercício 1
idade_total = df['Idade'].sum()
qnt_idade = df['Idade'].count()
idade_med = idade_total / qnt_idade
print(idade_med.round(2))
```

.sum() -> soma os valores da coluna 
.count() -> conta os valores não nulos da coluna
.round() -> corta os números depois da vírgula

```python
#Exercício 2

med_ngest = df['ngest'].mean()
print(med_ngest)
```

.mean() -> calcula a média de uma coluna

```python
#Exercício 3
df['parasit'].value_counts().sum()
null = df['parasit'].isnull().sum()

porc_null = null /len(df) * 100
print(f'{porc_null.round(3)}%')
```

isnull() -> retorna true e false para cada linha, junta com o sum()

.value_counts().sum() -> soma os valores não nulos

```python
# Exercício 4
nmal3 = df[df['nmal'] > 3]
print(len(nmal3))
```

```python
#Exercício 5

bins = [0, 18, 24, 29, 34, 39,140]
labels = [ '≤19', '20–24','25–29', '30–34', '35–39', '≥40 anos']

faixa_etaria = pd.cut(df['Idade'], bins=bins, labels = labels)
print(df['Idade'].groupby(faixa_etaria).mean().round(2))
```

bins e labels são do pd.cut(), que corta o df de acordo com as faixas determinadas

depois, uso groupby() para separar os valores feitos com a variavel do df.cut, e uso mean() para somar as médias
