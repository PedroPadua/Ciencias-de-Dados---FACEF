Conjuntos
– Um conjunto em Python é uma coleção de itens únicos que
não seguem uma ordem específica (sem elementos repetidos)
– É uma estrutura de dados mutável
– Permite a eliminação de itens duplicados

```python
lista = [1,2,3,4,3,2,1,5,8,7]

num_n_repetidos = set(lista)
print(num_n_repetidos)
```

DICIONÁRIOS

É útil para consultar, incluir, alterar ou remover itens e seus
significados, ou informações associadas
– São indexados por chaves (keys) de um tipo imutável, que
utilizam a estrutura chave: valor (key: value)
– São criados utilizando chaves, todavia, na inserção de
valores, é feita por colchetes

```python
dic_frutas = {
    'banana': 5.5,
    'Maçã' : 3.5,
    'Pera' : 6.5,
    'Uva' : 7.2
}

dic_frutas['Manga'] = 6.2
print(dic_frutas)
```

```python
print(f'Valor da Uva: {dic_frutas['Uva']}')
```

```python
# Construindo um dicionário com os valores sendo chaves

atletas = {}

nome_atleta = 'Início'
while nome_atleta !='':
    nome_atleta = input('Nome atleta:')

    if nome_atleta !='':
        saltos = []
        for i in range(0,2):
            saltos.append(float(input(f'{i+1}º Salto: '))) # Guarda os saltos do nome
        atletas[nome_atleta] = saltos # Atribui o salto ao nome 
```

```python
for nome_atleta, saltos in atletas.items():
    print(f'Média dos saltos do {nome_atleta}: {(sum(saltos)/2)}')
```

```python
#FOR em dicionários
dicionario = {
    'Pedro': 9,
    'Lucas': 4,
    'João': 2
}
for chave, valor in dicionario.items():
    print(f'Aluno {chave} tem a nota: {valor}')
```
